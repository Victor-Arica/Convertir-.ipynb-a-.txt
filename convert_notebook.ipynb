```python
import json

# Ruta del archivo .ipynb
ipynb_file_path = 'Matrices.ipynb'
txt_output_path = 'pregunta3.txt'

# Cargar el archivo .ipynb como JSON
with open(ipynb_file_path, 'r', encoding='utf-8') as file:
    notebook_content = json.load(file)

# Extraer el contenido de texto (de las celdas de tipo 'code' o 'markdown')
text_content = []
for cell in notebook_content['cells']:
    if cell['cell_type'] == 'markdown' or cell['cell_type'] == 'code':
        text_content.append("".join(cell['source']))

# Guardar el contenido extraído en un archivo .txt
with open(txt_output_path, 'w', encoding='utf-8') as txt_file:
    txt_file.write("\n\n".join(text_content))

print(f"Archivo convertido y guardado en {txt_output_path}")
```
